```python
import json
import spatialrust as sr
from spatialrust_jupyter import ViewerWidget
state = sr.ViewerState(640, 480)
widget = ViewerWidget(state, viewer_url='https://viewer.example.test/spatialrust/widget_embed.html')
widget.apply_input({'kind': 'zoom', 'delta': 1.0})
envelope = widget.transport_envelope()
assert envelope['transport_version'] == 1
assert envelope['state']['revision'] == 1
assert envelope['state']['viewer']['viewport'] == {'width': 640, 'height': 480}
print('SpatialRust Jupyter viewer smoke: PASS')
```
